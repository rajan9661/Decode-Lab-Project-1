2. WeekendOrder <br>  
Identify whether an order was placed on a weekend.   

```python
df['WeekendOrder'] = df['Date'].dt.dayofweek  
```

```python
df['WeekendOrder'] = df['WeekendOrder'].apply(
    lambda x: 1 if x >= 5 else 0
) 
```

WeekendOrder column means => <br> 
1 means weekend day (Saturady or Sunday) <br> 
0 means working day (Monday to Friday) 

```python
df.groupby('WeekendOrder')['OrderID'].nunique()  
```

```python
(df[df['WeekendOrder']==1]['OrderID'].count()/df.shape[0])*100   
```

Conclusion no 2 => 
Approx 30 percent order of total order  placed in weekend day 

3 => on the basis of unit price 

```python
def price_category(price):
    if price < 100:
        return 'Low'
    elif price < 500:
        return 'Medium'
    else:
        return 'High'

df['PriceCategory'] = df['UnitPrice'].apply(price_category)    
```
